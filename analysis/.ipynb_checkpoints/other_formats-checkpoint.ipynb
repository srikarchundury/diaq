```python
DATA_DIR = "../../data/"
import termcolor
def matrix_print01(nparray):
    for x in nparray:
        for y in x:
            a = int(y.real != 0.0 or y.imag != 0.0)
#             print(y.real, y.imag, a)
            if a != 0:
                print(termcolor.colored("*", "red") ,end=" ", sep="")
            else:
                print(a, end=" ", sep="")
        print()
import math

def convert_size(size_bytes):
   if size_bytes == 0:
       return "0B"
   size_name = ("B", "KB", "MB", "GB", "TB", "PB", "EB", "ZB", "YB")
   i = int(math.floor(math.log(size_bytes, 1024)))
   p = math.pow(1024, i)
   s = round(size_bytes / p, 2)
   return "%s %s" % (s, size_name[i])
```

```python
import os
import numpy as np
from scipy import sparse
def read_from_dir(direc="benchmarks/ghz/4", version="dia"):
    lm = []
    directory = DATA_DIR + direc
    for filename in os.listdir(directory):
        each_m = np.load(directory + "/" + filename)["arr_0"]
        lm.append(each_m)
    if version == "dia":
        lm = [sparse.dia_matrix(x) for x in lm]
    elif version == "csc":
        lm = [sparse.csc_matrix(x) for x in lm]
    elif version == "csr":
        lm = [sparse.csr_matrix(x) for x in lm]
    elif version == "coo":
        lm = [sparse.coo_matrix(x) for x in lm]
    elif version == "bsr":
        lm = [sparse.bsr_matrix(x) for x in lm]
    return lm
```

```python
chain_coo = read_from_dir(version="coo", direc="benchmarks/ham/13")
print("length: ", len(chain_coo))
print(type(chain_coo[0]))

chain_csr = read_from_dir(version="csr", direc="benchmarks/ham/13")
print("length: ", len(chain_csr))
print(type(chain_csr[0]))

chain_bsr = read_from_dir(version="bsr", direc="benchmarks/ham/13")
print("length: ", len(chain_bsr))
print(type(chain_bsr[0]))

chain_csc = read_from_dir(version="csc", direc="benchmarks/ham/13")
print("length: ", len(chain_csc))
print(type(chain_csc[0]))

chain_dia = read_from_dir(version="dia", direc="benchmarks/ham/13")
print("length: ", len(chain_dia))
print(type(chain_dia[0]))

chain_dense = read_from_dir(version="dense", direc="benchmarks/ham/13")
print("length: ", len(chain_dense))
print(type(chain_dense[0]))
```

```python
import time
from pympler import asizeof
def run(chain):
    mem = convert_size(asizeof.asizeof(chain))
    start = time.time()
#     print("matrix 0 - ", type(chain[0]))
#     print("matrix 1 - ", type(chain[1]))
    res = chain[1] * chain[0]
#     print("1 times 0 - ", type(res))
    for i in range(2,len(chain)):
#         print("matrix ", i, " - ", type(chain[i]))
        res = chain[i] * res
    print(type(res))
#     ans = res.todense()
#     ans = np.squeeze(np.asarray(ans))
#     print(type(ans))
#     matrix_print01(ans)
#     print(res)
    end = time.time()
    return end-start, mem
```
