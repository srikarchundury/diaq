```python
DATA_DIR = "../../data/"
import termcolor
def matrix_print01(nparray):
    for x in nparray:
        for y in x:
            a = int(y.real != 0.0 or y.imag != 0.0)
#             print(y.real, y.imag, a)
            if a != 0:
                print(termcolor.colored("*", "red") ,end=" ", sep="")
            else:
                print(a, end=" ", sep="")
        print()
```

```python
import os
import numpy as np
from scipy import sparse
def read_from_dir(direc="benchmarks/ghz/4", version="dia"):
    lm = []
    directory = DATA_DIR + direc
    for filename in os.listdir(directory):
        each_m = np.load(directory + "/" + filename)["arr_0"]
        lm.append(each_m)
    if version == "dia":
        lm = [sparse.dia_matrix(x) for x in lm]
    elif version == "csc":
        lm = [sparse.csc_matrix(x) for x in lm]
    elif version == "csr":
        lm = [sparse.csr_matrix(x) for x in lm]
    elif version == "coo":
        lm = [sparse.coo_matrix(x) for x in lm]
    elif version == "bsr":
        lm = [sparse.bsr_matrix(x) for x in lm]
    return lm
```

```python
chain = read_from_dir(version="bsr", direc="benchmarks/ghz/5")
print("length: ", len(chain))
print(type(chain[0]))
```

```python
def get_adds_muls(chain):
    adds = 0
    muls = 0
    print("matrix 0 - ", type(chain[0]))
    print("matrix 1 - ", type(chain[1]))
    res = chain[1] * chain[0]
    print("1 times 0 - ", type(res))
    for i in range(len(chain[2:])):
        print("matrix i - ", type(chain[i]))
        res = chain[i] * res
    print(type(res))
    ans = res.todense()
    ans = np.squeeze(np.asarray(ans))
    print(type(ans))
    matrix_print01(ans)
    return adds, muls
get_adds_muls(chain)
```
